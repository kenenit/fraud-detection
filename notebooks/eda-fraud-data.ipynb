```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid")
pd.set_option('display.max_columns', None)
```

```python
fraud_df = pd.read_csv('../data/raw/Fraud_Data.csv')
ip_df = pd.read_csv('../data/raw/IpAddress_to_Country.csv')

print("Fraud_Data shape:", fraud_df.shape)
print("IpAddress_to_Country shape:", ip_df.shape)
fraud_df.head()
```

```python
print("=== Data Types & Nulls ===")
print(fraud_df.info())
print("\n=== Missing Values ===")
print(fraud_df.isnull().sum())
print("\n=== Duplicates ===")
print("Duplicate rows:", fraud_df.duplicated().sum())
```

```python
fraud_df['signup_time'] = pd.to_datetime(fraud_df['signup_time'])
fraud_df['purchase_time'] = pd.to_datetime(fraud_df['purchase_time'])

print("signup_time dtype:", fraud_df['signup_time'].dtype)
print("purchase_time dtype:", fraud_df['purchase_time'].dtype)
```

```python
class_counts = fraud_df['class'].value_counts()
class_pct = fraud_df['class'].value_counts(normalize=True) * 100

print("=== Class Distribution ===")
print(pd.DataFrame({'Count': class_counts, 'Percentage': class_pct.round(2)}))

plt.figure(figsize=(6, 4))
sns.countplot(x='class', data=fraud_df, palette=['steelblue', 'crimson'])
plt.title('Class Distribution (0=Legit, 1=Fraud)')
plt.xticks([0, 1], ['Legitimate', 'Fraud'])
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('../data/processed/class_distribution.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.histplot(fraud_df['purchase_value'], bins=50, ax=axes[0], color='steelblue')
axes[0].set_title('Purchase Value Distribution')

sns.histplot(fraud_df['age'], bins=30, ax=axes[1], color='steelblue')
axes[1].set_title('Age Distribution')

fraud_df['source'].value_counts().plot(kind='bar', ax=axes[2], color='steelblue')
axes[2].set_title('Traffic Source')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('../data/processed/univariate_analysis.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Purchase value by class
sns.boxplot(x='class', y='purchase_value', data=fraud_df,
            palette=['steelblue', 'crimson'], ax=axes[0])
axes[0].set_title('Purchase Value by Class')
axes[0].set_xticklabels(['Legitimate', 'Fraud'])

# Age by class
sns.boxplot(x='class', y='age', data=fraud_df,
            palette=['steelblue', 'crimson'], ax=axes[1])
axes[1].set_title('Age by Class')
axes[1].set_xticklabels(['Legitimate', 'Fraud'])

# Fraud rate by source
fraud_by_source = fraud_df.groupby('source')['class'].mean().sort_values(ascending=False)
fraud_by_source.plot(kind='bar', ax=axes[2], color='crimson')
axes[2].set_title('Fraud Rate by Source')
axes[2].set_ylabel('Fraud Rate')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('../data/processed/bivariate_analysis.png', dpi=150)
plt.show()
```
